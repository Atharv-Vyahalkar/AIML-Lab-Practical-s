```python
graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'D': 5, 'E': 10},
    'C': {'F': 2},
    'D': {'G': 3},
    'E': {'G': 2},
    'F': {'G': 1}
}

heuristic = {
    'A': 11,
    'B': 7,
    'C': 5,
    'D': 3,
    'E': 2,
    'F': 1,
    'G': 0
}

def a_star(start, goal):

    random_list = [start]
    cost = {start: 0}
    parent = {start: None}

    while random_list:

        current = random_list[0]

        for node in random_list:
            if cost[node] + heuristic[node] < cost[current] + heuristic[current]:
                current = node

        if current == goal:

            path = []
            node = goal

            while node is not None:
                path.insert(0, node)
                node = parent[node]

            return path, cost[current]

        random_list.remove(current)

        for neighbor in graph[current]:

            new_cost = cost[current] + graph[current][neighbor]

            if neighbor not in cost or new_cost < cost[neighbor]:

                cost[neighbor] = new_cost
                parent[neighbor] = current

                if neighbor not in random_list:
                    random_list.append(neighbor)


path, total_cost = a_star('A', 'G')

print("Path:", path)
print("Cost:", total_cost)
```
