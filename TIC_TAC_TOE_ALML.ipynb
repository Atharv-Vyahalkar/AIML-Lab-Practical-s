```python
import random


board = [' ' for _ in range(9)]

def print_board(board):
  row1 = f'| {board[0]} | {board[1]} | {board[2]} |'
  row2 = f'| {board[3]} | {board[4]} | {board[5]} |'
  row3 = f'| {board[6]} | {board[7]} | {board[8]} |'
  print()
  print(row1)
  print(row2)
  print(row3)
  print()

def player_move(board):
  while True:
    try:
      position = int(input("Enter your move (0-8) to place 'X': "))
      if 0 <= position <= 8 and board[position] == ' ':
        board[position] = 'X'
        break
      else:
        print("Invalid move. That spot is already taken or out of bounds. Try again.")
    except ValueError:
      print("Invalid input. Please enter a number between 0 and 8.")

def computer_move(board):

  for i in range(9):
    if board[i] == ' ':
      board[i] = 'O'
      if check_win(board, 'O'):
        print(f"Computer chose position {i} for 'O' to win!")
        return
      board[i] = ' '


  for i in range(9):
    if board[i] == ' ':
      board[i] = 'X'
      if check_win(board, 'X'):
        board[i] = 'O'
        print(f"Computer chose position {i} for 'O' to block your win.")
        return
      board[i] = ' '


  if board[4] == ' ':
    board[4] = 'O'
    print(f"Computer chose position 4 for 'O' (center).")
    return


  corners = [0, 2, 6, 8]
  available_corners = [c for c in corners if board[c] == ' ']
  if available_corners:
    position = random.choice(available_corners)
    board[position] = 'O'
    print(f"Computer chose position {position} for 'O' (corner).")
    return


  sides = [1, 3, 5, 7]
  available_sides = [s for s in sides if board[s] == ' ']
  if available_sides:
    position = random.choice(available_sides)
    board[position] = 'O'
    print(f"Computer chose position {position} for 'O' (side).")
    return


  available_moves = [i for i, spot in enumerate(board) if spot == ' ']
  if available_moves:
    position = random.choice(available_moves)
    board[position] = 'O'
    print(f"Computer chose position {position} for 'O' randomly as fallback.")

def check_win(board, player):

  for i in range(0, 9, 3):
    if board[i] == board[i+1] == board[i+2] == player:
      return True

  for i in range(3):
    if board[i] == board[i+3] == board[i+6] == player:
      return True

  if board[0] == board[4] == board[8] == player:
    return True
  if board[2] == board[4] == board[6] == player:
    return True
  return False

def check_draw(board):
  return ' ' not in board


print("Tic-Tac-Toe Game")
print("X AS USER AND O AS COMPUTER")


while True:
  print_board(board)
  player_move(board)

  if check_win(board, 'X'):
    print_board(board)
    print("Congratulations! You win!")
    break
  if check_draw(board):
    print_board(board)
    print("It's a draw!")
    break


  computer_move(board)

  if check_win(board, 'O'):
    print_board(board)
    print("Computer wins! Better luck next time.")
    break
  if check_draw(board):
    print_board(board)
    print("It's a draw!")
    break
```
